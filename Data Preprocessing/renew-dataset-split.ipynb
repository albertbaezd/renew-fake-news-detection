```python
import pandas as pd
from ftfy import fix_text

true_path = 'Cleaned dataset/renew/renew true cleaned.csv'
false_path = 'Cleaned dataset/renew/renew false cleaned.csv'

df_true = pd.read_csv(true_path, encoding='ISO-8859-1', encoding_errors='replace')
df_false = pd.read_csv(false_path, encoding='ISO-8859-1', encoding_errors='replace')

def safe_fix(text):
    if isinstance(text, str):
        return fix_text(text)
    return ""

for col in ['title', 'text']:
    if col in df_true.columns:
        df_true[col] = df_true[col].apply(safe_fix)

for col in ['title', 'text', 'News_body']:
    if col in df_false.columns:
        df_false[col] = df_false[col].apply(safe_fix)

print("True Dataset (cleaned):")
print(df_true[['title', 'text']].head(2))

print("\nFalse Dataset (cleaned):")
print(df_false[['title', 'text', 'News_body']].head(2))
```

```python
import os

output_dir = 'Cleaned dataset/renew/parsed'
os.makedirs(output_dir, exist_ok=True)

true_output_path = os.path.join(output_dir, 'renew true cleaned parsed.csv')
false_output_path = os.path.join(output_dir, 'renew false cleaned parsed.csv')

df_true.to_csv(true_output_path, index=False)
df_false.to_csv(false_output_path, index=False)

print("Cleaned CSVs have been saved to:")
print(true_output_path)
print(false_output_path)
```

```python
unique_sources = df_false['source'].dropna().unique()

print("Unique values in the 'source' column:")
for source in unique_sources:
    print(f"- {source}")
```

First step: Taking the LIAR dataset and saving a custom version of it with numeric values for the 'label' column.


```python
import pandas as pd

liar_train_path = 'LIAR dataset/train modified.tsv'
liar_df_train = pd.read_csv(liar_train_path, sep='\t', header=0)

liar_df_train.head()
```

```python
liar_true_labels = ['TRUE', 'mostly-true']

liar_df_train['label'] = liar_df_train['label'].apply(
    lambda x: 1 if x.strip() in liar_true_labels else 0
)

liar_df_train['label'].value_counts()
```

```python
liar_df_cleaned_train = liar_df_train[['statement', 'label']].dropna()

liar_df_cleaned_train.head()
```

```python
liar_df_cleaned_train.to_csv('LIAR dataset/cleaned for training/cleaned_train.csv', index=False)
```

Creating a Custom dataset with both LIAR and ReNew for testing and saving it a CSV.


```python
import pandas as pd

path1 = "./LIAR dataset/cleaned for training/cleaned_train.csv"
path2 = "./Cleaned dataset/renew/combined/combined renew train.csv"

df1 = pd.read_csv(path1)
df2 = pd.read_csv(path2)

print(df1.head())
print(df2.head())
```

```python

combined_df = pd.concat([df1, df2], ignore_index=True)

combined_df = combined_df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Class distribution:")
print(combined_df['label'].value_counts())
```

```python
import os

output_dir = "./LIAR dataset/combined liar and renew"
output_file = os.path.join(output_dir, "liar and renew training.csv")

os.makedirs(output_dir, exist_ok=True)

combined_df.to_csv(output_file, index=False)

print(f"Saved combined dataset to: {output_file}")
```
