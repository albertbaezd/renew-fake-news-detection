```python
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
df = pd.read_csv('./data/test_modified.tsv', sep='\t')

label_map = {
    "true": 1,
    "mostly-true": 1,
    "half-true": 0,
    "barely-true": 0,
    "false": 0,
    "pants-on-fire": 0
}
df["label"] = df["label"].map(label_map)
df = df.dropna(subset=["label"])
```

```python
tokenizer = AutoTokenizer.from_pretrained("./model")
model = AutoModelForSequenceClassification.from_pretrained("./model").to(device)
model.eval()

encodings = tokenizer(
    list(df["statement"]),
    truncation=True,
    padding=True,
    max_length=128,
    return_tensors="pt"
)
labels = torch.tensor(df["label"].tolist())
```

```python
dataset = TensorDataset(encodings["input_ids"], encodings["attention_mask"], labels)
loader = DataLoader(dataset, batch_size=32)

all_preds = []
all_labels = []
```

```python
for input_ids, attention_mask, batch_labels in tqdm(loader, desc="Batches"):
    input_ids = input_ids.to(device)
    attention_mask = attention_mask.to(device)
    batch_labels = batch_labels.to(device)

    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        preds = torch.argmax(outputs.logits, dim=-1)

    all_preds.extend(preds.cpu().tolist())
    all_labels.extend(batch_labels.cpu().tolist())
```

### Test with LIAR

```python
y_true = all_labels
y_pred = all_preds

accuracy = accuracy_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
specificity = tn / (tn + fp)

print("\nEvaluation Results:")
print(f"Accuracy:    {accuracy:.4f}")
print(f"Precision:   {precision:.4f}")
print(f"Recall:      {recall:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"F1 Score:    {f1:.4f}")
```

### Testing in ReNew Dataset

```python
df2 = pd.read_csv('./data/renew_combined_test.csv')

df2["label"] = df2["label"]
df2 = df2.dropna(subset=["label"])

encodings2 = tokenizer(
    list(df2["quotes"]),
    truncation=True,
    padding=True,
    max_length=128,
    return_tensors="pt"
)
labels2 = torch.tensor(df2["label"].tolist())

```

```python
dataset2 = TensorDataset(encodings2["input_ids"], encodings2["attention_mask"], labels2)
loader2 = DataLoader(dataset2, batch_size=32)

all_preds2 = []
all_labels2 = []

print("\nEvaluating on renew_combined_test.csv...")
for input_ids, attention_mask, batch_labels in tqdm(loader2, desc="Batches"):
    input_ids = input_ids.to(device)
    attention_mask = attention_mask.to(device)
    batch_labels = batch_labels.to(device)

    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        preds = torch.argmax(outputs.logits, dim=-1)

    all_preds2.extend(preds.cpu().tolist())
    all_labels2.extend(batch_labels.cpu().tolist())
```

```python
y_true2 = all_labels2
y_pred2 = all_preds2

accuracy2 = accuracy_score(y_true2, y_pred2)
recall2 = recall_score(y_true2, y_pred2)
precision2 = precision_score(y_true2, y_pred2)
f1_2 = f1_score(y_true2, y_pred2)
tn2, fp2, fn2, tp2 = confusion_matrix(y_true2, y_pred2).ravel()
specificity2 = tn2 / (tn2 + fp2)

print("\nResults on renew_combined_test.csv:")
print(f"Accuracy:    {accuracy2:.4f}")
print(f"Precision:   {precision2:.4f}")
print(f"Recall:      {recall2:.4f}")
print(f"Specificity: {specificity2:.4f}")
print(f"F1 Score:    {f1_2:.4f}")
```
