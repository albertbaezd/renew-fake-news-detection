Default model performance on the whole ReNew dataset


```python
import pandas as pd
import torch
from datasets import Dataset
from transformers import pipeline
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

renew_dataset_path = './ReNew dataset/combined renew dataset.csv'
renew_df = pd.read_csv(renew_dataset_path)

renew_hf_dataset = Dataset.from_pandas(renew_df)

renew_statements = renew_hf_dataset['statement']
renew_true_labels = renew_hf_dataset['label']

fake_news_model_path = "./jy46604790_fake_news_bert_detect"
fake_news_classifier = pipeline(
    "text-classification",
    model=fake_news_model_path,
    tokenizer=fake_news_model_path,
    device=0 if torch.cuda.is_available() else -1
)

renew_predictions_raw = fake_news_classifier(
    renew_statements,
    truncation=True,
    batch_size=32
)

renew_predicted_labels = [1 if pred['label'] == 'LABEL_1' else 0 for pred in renew_predictions_raw]

renew_accuracy = accuracy_score(renew_true_labels, renew_predicted_labels)
renew_f1_score = f1_score(renew_true_labels, renew_predicted_labels)
renew_confusion_matrix = confusion_matrix(renew_true_labels, renew_predicted_labels)

print(f"Accuracy on Renew dataset: {renew_accuracy:.4f}")
print(f"F1 Score on Renew dataset: {renew_f1_score:.4f}")
print("Confusion Matrix (Renew dataset):")
print(renew_confusion_matrix)
```

Fine tuning with ReNew training to test performance over ReNew testing


```python
import pandas as pd
from datasets import Dataset

combined_renew_train_csv_path = './ReNew dataset/combined renew train.csv'
combined_renew_train_df = pd.read_csv(combined_renew_train_csv_path)

combined_renew_train_hf_dataset = Dataset.from_pandas(combined_renew_train_df)

combined_renew_train_hf_dataset
```

```python
from transformers import RobertaTokenizer, RobertaForSequenceClassification

model_name = "jy46604790/Fake-News-Bert-Detect"
tokenizer = RobertaTokenizer.from_pretrained(model_name)
model = RobertaForSequenceClassification.from_pretrained(model_name, num_labels=2)

model.config, tokenizer
```

```python
def tokenize_function(examples):
    return tokenizer(examples['statement'], padding="max_length", truncation=True)

combined_renew_train_hf_dataset = combined_renew_train_hf_dataset.map(tokenize_function, batched=True)

combined_renew_train_hf_dataset
```

```python
train_dataset, val_dataset = combined_renew_train_hf_dataset.train_test_split(test_size=0.1).values()

len(train_dataset), len(val_dataset)
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./bert_fakenews_finetuned",
    evaluation_strategy="no",
    save_strategy="no",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=64,
    num_train_epochs=2,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=50,
    save_total_limit=1,
    load_best_model_at_end=True,
    metric_for_best_model="eval_f1",
    greater_is_better=True,
    fp16=True,
    report_to="none"
)
```

```python
import torch
print("Using GPU:", torch.cuda.is_available())
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    tokenizer=tokenizer,
)

trainer.train()
```

```python
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

predictions = trainer.predict(val_dataset)
predicted_labels = predictions.predictions.argmax(axis=-1)
ground_truth_labels = predictions.label_ids

accuracy = accuracy_score(ground_truth_labels, predicted_labels)

f1 = f1_score(ground_truth_labels, predicted_labels)

conf_matrix = confusion_matrix(ground_truth_labels, predicted_labels)

print(f"Accuracy: {accuracy:.4f}")
print(f"F1 Score: {f1:.4f}")
print("Confusion Matrix:")
print(conf_matrix)
```

```python
output_dir = "./fine tuned models/jy46_fine_tuned_fake_news_bert"

model.save_pretrained(output_dir)
tokenizer.save_pretrained(output_dir)

print(f"Model and tokenizer saved to {output_dir}")
```

Testing performance over the original politifact dataset, using the original (default) moodel


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

folder_path = "./jy46604790_fake_news_bert_detect"

politifact_tokenizer = AutoTokenizer.from_pretrained(folder_path)
politifact_model = AutoModelForSequenceClassification.from_pretrained(folder_path)
```

```python
politifact_fake_df = pd.read_csv("./Politifact dataset/politifact kaggle fake.csv")
politifact_true_df = pd.read_csv("./Politifact dataset/politifact kaggle true.csv")

print("Fake News Samples:")
print(politifact_fake_df.head())

print("\nTrue News Samples:")
print(politifact_true_df.head())
```

```python
politifact_fake_df["label"] = 0
politifact_true_df["label"] = 1

politifact_combined_df = pd.concat([politifact_fake_df, politifact_true_df], ignore_index=True)

politifact_combined_df = politifact_combined_df.sample(frac=1, random_state=42).reset_index(drop=True)

print(politifact_combined_df.head())
```

```python

train_df, test_df = train_test_split(
    politifact_combined_df, 
    test_size=0.2, 
    random_state=42,
    stratify=politifact_combined_df["label"]
)

print(f"Training samples: {len(train_df)}")
print(f"Testing samples: {len(test_df)}")
```

```python
import os

os.makedirs("./Politifact dataset/merged", exist_ok=True)

train_file_path = "./Politifact dataset/merged/merged_politifact_train.csv"
test_file_path = "./Politifact dataset/merged/merged_politifact_test.csv"

train_df.to_csv(train_file_path, index=False)
test_df.to_csv(test_file_path, index=False)

print(f"Training dataset saved to: {train_file_path}")
print(f"Testing dataset saved to: {test_file_path}")
```

```python
politifact_copy_df = test_df.copy()

politifact_copy_df = politifact_copy_df.rename(columns={"text": "statement"})

politifact_statements = politifact_copy_df['statement'].tolist()
politifact_true_labels = politifact_copy_df['label'].tolist()

print(politifact_copy_df.head())
```

```python
from transformers import pipeline

fake_news_model_path = "./jy46604790_fake_news_bert_detect"

politifact_fake_news_classifier = pipeline(
    "text-classification",
    model=fake_news_model_path,
    tokenizer=fake_news_model_path,
    device=0 if torch.cuda.is_available() else -1
)
```

```python
politifact_predictions_raw = politifact_fake_news_classifier(
    politifact_statements,
    truncation=True,
    batch_size=32
)

politifact_predicted_labels = [1 if pred['label'] == 'LABEL_1' else 0 for pred in politifact_predictions_raw]
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, confusion_matrix, f1_score

politifact_accuracy = accuracy_score(politifact_true_labels, politifact_predicted_labels)
politifact_f1_score = f1_score(politifact_true_labels, politifact_predicted_labels)
politifact_recall = recall_score(politifact_true_labels, politifact_predicted_labels)
politifact_precision = precision_score(politifact_true_labels, politifact_predicted_labels)

politifact_conf_matrix = confusion_matrix(politifact_true_labels, politifact_predicted_labels)

tn, fp, fn, tp = politifact_conf_matrix.ravel()
politifact_specificity = tn / (tn + fp)

print(f"Accuracy on Politifact dataset: {politifact_accuracy:.4f}")
print(f"Recall on Politifact dataset: {politifact_recall:.4f}")
print(f"Specificity on Politifact dataset: {politifact_specificity:.4f}")
print(f"Precision on Politifact dataset: {politifact_precision:.4f}")
print(f"F1 Score on Politifact dataset: {politifact_f1_score:.4f}")
print("Confusion Matrix (Politifact dataset):")
print(politifact_conf_matrix)
```
